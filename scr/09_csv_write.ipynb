```python
import csv
```

```python
makes = ['Toyota', 'Ford', 'BMW']
models = ['RAV4', 'F150', 'X6']
max_speed = [220, 200, 260]
powers = [195, 400, 360]
```

```python
cars_list = list(zip(makes, models, max_speed, powers))
cars_list
```

```python
with open('../data/Cars.csv', 'w', newline='') as f:
  writer = csv.writer(f)
  writer.writerow(['Make', 'Model', 'Max Speed', 'Powers'])
  writer.writerows(cars_list)
```

```python
with open('../data/Cars_line.csv', 'w') as f:
  writer = csv.writer(f)
  writer.writerow(['Make', 'Model', 'Max Speed', 'Powers'])
  writer.writerows(cars_list)
```
